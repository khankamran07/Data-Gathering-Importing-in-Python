### 2.1) Opening csv file from URL

```python
# To open a CSV file from a URL, you just need to remember this code and paste the link in place of the URL.

import requests
from io import StringIO

url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

```python
# Here, the values are not separated by commas; they are separated by tabs.
# Also, there are no column names.
# To solve this, check the code snippet below.

pd.read_csv('movie_titles_metadata.tsv')
```

### 2.2) Sep Parameter

```python
# sep means separator, and its default value is a comma (sep=',').
# In the file below, we have a TSV file, where the values are not separated by commas. 
# They are separated by tabs, so we use sep='\t'.

# If the file does not have column names, you can pass a list containing the column names, and Pandas will assign those names to the columns.

pd.read_csv('movie_titles_metadata.tsv',sep='\t',names=['sno','movies','release_year','rating','votes','genres'])
```

### 2.3) Index_col parameter

```python
# Here, the first index is useless, so we want to make enrollee_id the index instead.

pd.read_csv('aug_train.csv')
```

```python
# If there is a column that is not useful, for example, if you want to make enrollee_id the index, you can convert that column into the index.

pd.read_csv('aug_train.csv',index_col='enrollee_id')
```

### 2.4) Header parameter

```python
# Here,the first row contains the column names. We want to convert the first row into column names and start the actual data from the second row.
# You can see this in the code snippet below.

pd.read_csv('test.csv')
```

```python
# The header parameter is used while reading data to specify which row should be used as the column names (headers).

pd.read_csv('test.csv',header=1)
```

### 2.5) use_cols parameter

```python
# Sometimes, we don't need to use all the columns. If I need only a few specific columns, I can provide their names in usecols.
# For example:usecols=['enrollee_id', 'gender', 'education_level']
#Only these columns will be included, and the remaining columns will be removed.

pd.read_csv('aug_train.csv',usecols=['enrollee_id','gender','education_level'])
```

### 2.6) Squeeze parameters

```python
# Sometimes, you only need a single column. If you use squeeze="column", it converts the result from a DataFrame into a Series. 

pd.read_csv('aug_train.csv', usecols=['gender']).squeeze("columns")
```

### 2.7) Skiprows/nrows Parameter

```python
# Skiprows
# Whatever values you provide here, the corresponding rows will be skipped.
# For eg-if I want to remove the rows with values 0 and 5, I can specify 0 and 5, and those rows will be skipped.


pd.read_csv('aug_train.csv',skiprows=[1,2])
```

```python
# nrows
# Suppose the CSV file has 10,000 rows. If you set `n=100`, it will read only the first 100 rows.
# The advantage is that if you are working with a very large dataset, such as 2 million rows, and your computer does not have enough memory or storage,
# you can load the data in **smaller chunks** and perform your analysis.


pd.read_csv('aug_train.csv',nrows=100)
```

### 2.8) Encoding parameter

```python
# The `encoding` parameter is used to correctly read the characters and text in a file.
# Here, encoding='utf-8 tells Pandas TO Read the text in this file according to UTF-8 encoding.


pd.read_csv('zomato.csv',encoding='latin-1')
```

### 2.9) Skip bad lines

```python
# In Pandas, the on_bad_lines parameter is used to decide what to do when there is a problematic or incorrect row in a CSV file.

pd.read_csv('BX-Books.csv', sep=';', encoding="latin-1",on_bad_lines='skip')
```

### 2.10) dtypes parameter

```python
# Here, the target column at the end of this CSV file is in float format. If you don't want it as a float and want it to be an integer
# you can use `dtype` to change its data type.


pd.read_csv('aug_train.csv')
```

```python
# The `dtype` parameter is used to change the data type, such as converting float to int or int to float.


pd.read_csv('aug_train.csv',dtype={'target':int}).info()
```

### 2.11) Handling Dates

```python
# Whenever you use a CSV file, the values are treated as **strings by default**.
# If a date is stored as a string, you cannot use date-related functionality, such as finding the **month, day, or year**.
# To keep the values in **date-time format**, you need to use this.


pd.read_csv('IPL Matches 2008-2020.csv',parse_dates=['date']).info()
```

### 2.12) Convertors

```python
def rename(name):
    if name == "Royal Challengers Bangalore":
        return "RCB"
    else:
        return name
```

```python
rename("Royal Challengers Bangalore")
```

```python
# It is used to transform names. For example, ‘Royal Challengers Bangalore’ can be transformed into ‘RCB’.

pd.read_csv('IPL Matches 2008-2020.csv',converters={'team1':rename})
```

### 2.13) na_values parameter

```python
# This makes the value NaN when you do this

pd.read_csv('aug_train.csv',na_values=['Male'])
```

### 2.14) Loading a huge dataset in chunks

```python
dfs = pd.read_csv('aug_train.csv',chunksize=5000)
```

```python
for chunks in dfs:
    print(chunks.shape)
```

## 6.Import Data from SQL 

```python
import mysql.connector
```

```python
# In this, you pass four things:

# First you provide the IP address of your database server. If the database is running on your own machine, you can set the host value to localhost.
# Second, you provide the username of that machine. When we work with XAMPP, our username is usually root.
# Third, you provide the password. In our case, the password is empty.
# Fourth, you need to specify which database is present on your database server. We have saved a database named world in XAMPP,
# so we provide world as the database name.

conn = mysql.connector.connect(host = 'localhost',user='root',password = '',database='world') 
```

```python
pd.read_sql_query("SELECT * FROM city",conn)
```

```python
pd.read_sql_query("SELECT * FROM city WHERE CountryCode LIKE 'AFG'",conn)
```

```python
# Let’s say I need data about countries.
 
pd.read_sql_query("SELECT * FROM country",conn)
```
